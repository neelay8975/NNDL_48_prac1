```python
# array addition
import numpy as np
v1=np.array([95,79,10,94,42])
v2=np.array([89,75,83,23,74])
print("Vector1: ",v1)
print("Vector2: ",v2)
print("Addition: ",v1+v2)
```

```python
import numpy as np
m1=np.array([[95,79],
             [34,98]])
m2=np.array([[56,76],
             [34,12]])
print("matrix1: ",m1)
print("matrix2: ",m2)
print("multiplication: ",np.matmul(m1,m2))
```
